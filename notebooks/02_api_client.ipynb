```python
import requests
import pandas as pd

API_URL = "http://127.0.0.1:8000"

print("API client ready")
```

```python
response = requests.get(
    f"{API_URL}/health"
)

print("Status code:", response.status_code)
print("Response:")
print(response.json())
```

```python
response = requests.get(
    f"{API_URL}/model/info"
)

print("Status code:", response.status_code)

model_info = response.json()

print("Model type:", model_info["model_type"])
print("Model loaded:", model_info["model_loaded"])
print("Features:", model_info["features"])
print("Target:", model_info["target"])
```

```python
property_data = {
    "bed": 3,
    "bath": 2,
    "house_size": 1800,
    "acre_lot": 0.25,
    "state": "California",
    "city": "Los Angeles",
    "zip_code": "90001"
}

response = requests.post(
    f"{API_URL}/predict",
    json=property_data
)

print("Status code:", response.status_code)
print("Response:", response.json())
```
